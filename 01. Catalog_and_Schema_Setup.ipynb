```python
%sql
CREATE CATALOG IF NOT EXISTS pii_demo;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS pii_demo.pii_source;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS pii_demo.pii_target;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS pii_demo.pii_source.raw_data;
```

```python
table_names = ['customers', 'employees', 'web_logs']
```

```python
for table in table_names:
    spark.sql(f"DROP TABLE IF EXISTS pii_demo.pii_source.{table}")
    spark.sql(f"""CREATE TABLE IF NOT EXISTS pii_demo.pii_source.{table}
                    USING DELTA
                    AS
                    SELECT * FROM read_files('/Volumes/pii_demo/pii_source/raw_data/{table}.csv',
                    format => 'csv',
                    header => true,
                    inferSchema => true)
                    """)
```
